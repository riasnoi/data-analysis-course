## Подготовка

```python
from pathlib import Path
import sys
import pandas as pd

print("Python:", sys.executable)
print("pandas:", pd.__version__)
print("Рабочая папка:", Path.cwd())
DATA = Path("../data")
assert DATA.is_dir(), "Рядом с папкой ноутбука должна находиться data/"
```

## Основной набор decisions.csv

```python
home_decisions = pd.read_csv(
    DATA / "decisions.csv", dtype={"lei": "string", "county_code": "string"}
)
home_outcomes = home_decisions["action_taken"]
home_n = len(home_decisions)
assert home_n > 0
assert home_outcomes.isna().sum() == 0
assert home_outcomes.isin([1, 2, 3]).sum() == home_n
print(home_decisions.shape)
home_outcomes.value_counts(dropna=False)
```

## H1. Доля отказов

```python
total_applications = len(home_outcomes)
rejections = (home_outcomes == 3).sum()
rejection_rate = (rejections / total_applications) * 100

print(f"Всего заявок (знаменатель): {total_applications}")
print(f"Отказов (числитель): {rejections}")
print(f"Доля отказов: {rejection_rate:.2f}%")

assert total_applications > 0
assert 0 <= rejection_rate <= 100
```

## H2. Одобрение без принятия предложения

```python
approvals_not_accepted = (home_outcomes == 2).sum()
total_approved = (home_outcomes == 1).sum() + (home_outcomes == 2).sum()

rate_h2 = (approvals_not_accepted / total_approved) * 100

print(f"Одобрено, но не принято (числитель): {approvals_not_accepted}")
print(f"Всего одобренных заявок (знаменатель): {total_approved}")
print(f"Доля от одобренных: {rate_h2:.2f}%")
```
